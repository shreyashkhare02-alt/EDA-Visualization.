Titanic Passenger Analysis
You are a data analyst studying passenger data from the Titanic ship. The dataset contains information about

```python
df=df.fillna(0)
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

Q1. Create a countplot to show the number of passengers by gender.

```python
sns.countplot(x='Sex', data=df,)
plt.title("Number of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

Q2. Create a bar chart to show the number of passengers in each passenger class using different colors

```python
df['Pclass'].value_counts().plot(kind='bar')
plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

Q3. Create a histogram to show the distribution of passenger age

```python
plt.hist(df['Age'])
plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
```

Q4. Create a boxplot to detect outliers in the fare column.

```python
sns.boxplot(x=df['Fare'], color='Pink')
plt.title("Boxplot of Passenger Fare")
plt.xlabel("Fare")
plt.show()
```

Q5. Create a scatterplot to

a. show the relationship between age and fare.




b. analyze the relationship between family size (SibSp + Parch) and fare

```python
#a
sns.scatterplot(x='Age', y='Fare', data=df, color='green')
plt.title("Relationship between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch']
sns.scatterplot(x='FamilySize', y='Fare', data=df, color='orange')
plt.title("Relationship between Family Size and Fare")
plt.xlabel("Family Size (SibSp + Parch)")
plt.ylabel("Fare")
plt.show()
```


Q6. Create a countplot to show survival count by gender

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

Q7. Create a bar chart to show survival by passenger class.

```python
sns.countplot(x='Pclass', hue='Survived', data=df, )
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

Q8. Create a heatmap to show correlation between numerical variables.

```python
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True)
plt.title("correlation Heatmap")
plt.show()
```

Q9. Identify which passenger class paid the highest fare on average

```python
import pandas as pd
avg_fare_by_class = df.groupby('Pclass')['Fare'].mean()
print(avg_fare_by_class)

avg_fare_by_class.plot(kind='bar')
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

Key Insights from Titanic EDA

---


**1.Gender Effect on Survival**

Women had a much higher survival rate
compared to men.

The evacuation policy of “women and children first” is clearly reflected in the data.

---



**2.Class Effect on Survival**

1st class passengers had the highest survival chances.

3rd class passengers had the lowest survival rate, showing socio-economic status strongly influenced survival.

---



**3.Age Distribution & Survival**

Most passengers were between 20–40 years old.

Children (especially under 10 years) had better survival chances compared to adults.

---



4.Fare & Survival Relationship
**bold text**
Passengers who paid higher fares (linked to 1st class) were more likely to survive.

Outliers with very high fares correspond to wealthy passengers who had better access to lifeboats.

---



**5.Family Size Impact**

Solo travelers and small families had better survival chances.

Very large families struggled, with lower survival rates, likely due to difficulties evacuating together.
